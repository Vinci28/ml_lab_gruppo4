```python
from pathlib import Path

from IPython.core.pylabtools import figsize
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
CSV_PATH = Path.cwd() / "Gruppo4_stud_1_N" / "Accelerometer.csv"
```

```python
data = np.genfromtxt(CSV_PATH, delimiter=",", names=True)
```

```python
seconds = data["seconds_elapsed"]
```

```python
x = data["x"]
```

```python
y = data["y"]
```

```python
z = data["z"]
```

```python
len(x)
```

```python
z[:5]
```

```python
fig, axes = plt.subplots(3, 1, sharex=True, figsize=(12, 8))

for ax, signal, label, color in zip(
    axes, (x, y, z), ("x", "y", "z"), ("tab:red", "tab:green", "tab:blue")
):
    ax.plot(seconds, signal, color=color, linewidth=0.8)
    ax.set_ylabel(f"acc {label} (m/s²)")
    ax.grid(True, alpha=0.3)

axes[-1].set_xlabel("tempo (s)")
fig.suptitle("Accelerometro")
fig.tight_layout()
plt.show()
```
